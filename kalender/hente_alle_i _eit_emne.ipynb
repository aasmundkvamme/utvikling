```python
import requests
import os
import pandas as pd
TOKEN_CANVAS = os.environ['TOKEN_CANVAS']

headers = {
    'Authorization': f'Bearer {TOKEN_CANVAS}',
}
params = {
    'per_page': 100,
    'context_codes[]': 'course_32013',
}
```

```python
url = f"https://hvl.instructure.com/api/v1/calendar_events"
response = requests.get(url, headers=headers, params=params)
data = response.json()
df = pd.json_normalize(data)
df
```
