```python
import pandas as pd
import requests
import os

TOKEN_CANVAS = os.environ['TOKEN_CANVAS']

headers = {
    'Authorization': f'Bearer {TOKEN_CANVAS}',
}
params = {
    'per_page': 100,
    'show_inherited': True
}
```

```python
data = pd.read_csv('provisioning_csv_26_Sep_2025_6316120250926-2472206-h4myvm.csv')
```

```python
data.info()
```

```python
oppgåveliste = []
for emne in data['canvas_course_id']:
    url = f'https://hvl.instructure.com/api/v1/courses/{emne}/assignments'
    response = requests.get(url, headers=headers, params=params)
    if response.status_code == 200:
        assignments = response.json()
        for a in assignments:
            oppgåveliste.append((emne, a['id'], a['name'], a['peer_reviews']))

        # fagfellevurderte = [a for a in assignments if 'peer_review' in a]
        # if len(fagfellevurderte) > 0:
        #     fagfelleliste.append((emne, len(fagfellevurderte)))
        # print(f'Emne {emne} har {len(fagfellevurderte)} fagfellevurderte oppgaver.')
```

```python
oppgåveliste_df = pd.DataFrame(oppgåveliste, columns=['canvas_course_id', 'assignment_id', 'assignment_name', 'peer_reviews'])
```
