Det nye i år er at eg har lært (?) at eg kan lese ei tekstfil med data i linjer inn i ei liste med `readlines()`:

```python
with open('input1_1.txt') as f:
    data = f.readlines()
test = ['L68','L30','R48','L5','R60','L55','L1','L99','R14','L82']
```

Så lager eg ein funksjon som finn neste verdi (modulo 100):

```python
def flytt(start, retning, steg):
    if retning == 'L':
        return (start - steg) % 100
    else:
        return (start + steg) % 100
```

Og til slutt går eg gjennom data og tel opp kor mange ganger vi stopper i null:

```python
stopp_i_null = 0
start = 50
for kommando in data:
    retning = kommando[0]
    steg = int(kommando[1:])
    start = flytt(start, retning, steg)
    # print(start)
    if start == 0:
        stopp_i_null += 1
print(f'Stopp i null: {stopp_i_null} ganger')
```

Neste steg er å finne kor mange ganger vi *passerer* 0, ikkje berre *stopper* i 0. Så då bør eg kanskje legge inn ein "skifte forteikn"-test: eg definerer `start` som plassen vi starter i, og `stopp` som plassen vi stopper i. Dersom dei har ulikt forteikn har vi passert 0.

Men så viser det seg at eg hoppa over ein viktig informasjon: "Be careful: if the dial were pointing at 50, a single rotation like R1000 would cause the dial to point at 0 ten times before returning back to 50!" 

Så eg må bygge om rutina mi, til å faktisk *telje* antal omdreiingar, eller rekne? 

Ved prøving og feiling får eg algoritmen
> Start i posisjon $n_i$ og gå $s$ steg i retning $d$ (der $d=1$ eller $d=-1$)
> 
> Rekn ut ny posisjon $n_{i+1} = (n+d\cdot s)\bmod 100$.
>
> Rekn ut antal "stopp ved $0$" ved $\displaystyle k = \left\lfloor\frac{|n_i+s|}{100}\right\rfloor$

Men her manglar eg litt: dersom eg går eitt steg 

```python
stopp_i_null = 0
start = 50
for kommando in data:
    retning = -1 if kommando[0]=='L' else +1
    steg = int(kommando[1:])
    k = abs(start + steg)//100
    stopp = (start + retning*steg) % 100
    print(f"{start} {' +' if retning > 0 else ' -'} {steg} - > {start + retning*steg}, {k} passeringar av 0")
    start = stopp
    stopp_i_null += k
print(f"Totalt {stopp_i_null} passeringar av 0")
```
