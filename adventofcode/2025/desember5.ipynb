```python
datasett = data
friske = set()
intervall = []
for t in datasett:
    if t.find('-') > 0:
        start = int(t.split('-')[0])
        slutt = int(t.split('-')[1])
        # print(start, slutt)
        intervall.append((start, slutt))
    else:
        if t != '':
            tal = int(t)
            for i in intervall:
                if i[0] <= tal <= i[1]:
                    # print(f"{i[0]} <= {tal} <= {i[1]}") 
                    friske.add(tal)
print(f"Det er {len(friske)} friske råvarer.")
```
