```python
sum = 0
nummer = 1
for bank in data:
    bank = bank[0:-1]
    lengde = len(bank)
    tal = sorted(set(bank), reverse=True)
    # print(tal)
    i = 0
    av = True
    while av:
        posmax = bank.index(tal[i])
        # print(tal, tal[i], posmax)
        if bank.index(tal[i]) < lengde - 1:   # - 1
            # print("Fant største")
            bankny = bank[posmax+1:]
            talny = sorted(set(bankny), reverse=True)
            # print(" ",talny)
            posny = bankny.index(talny[i])     # i+1
            # print(" ",talny[i], posny)
            joltage = str(tal[i])+str(talny[i])
            sum += int(joltage)
        else:
            print("Tar neste")
            posmax = bank.index(tal[i+1])   # tal[i+1]
            # print(tal[i+1], posmax)
            # print(" ",max(bank[posmax+1:]))
            joltage = str(tal[i+1])+str(max(bank[posmax+1:]))
            # print(joltage)
            sum += int(joltage)
        av = False
        print('*', nummer, joltage)
        nummer += 1
print(sum)
```
